#### You need an API key to use the API, but not safe to upload API keys to git (especially if we pay for services).  You should be able to get one from Vectara console.

```python
api_key = "" # decided not to check in to git.  Obtainable from Vectara console
```

```python
corpus_key = "SGU_Discourse_3"
```

```python
import requests
import json
import pathlib

url = f"https://api.vectara.io/v2/corpora/{corpus_key}/documents"

headers = {
  'Content-Type': 'application/json',
  'Accept': 'application/json',
  'x-api-key': f'{api_key}'
}
```

```python
def index_episode(episode_number):
    """Uploads the episode with the given episode number to Vectara.

    It relies on the json object for the episode already being in the json folder

    Args:
        episode_number: integer episode number
    """
    path = pathlib.Path(f"./ep_json/SGU_Episode_{episode_number}.json")
    payload = json.loads(path.read_text(encoding="utf-8"))

    response = requests.post(url, headers=headers, json=payload)
    print(response)
    print(response.text)
```

#### Upload Epidoses

```python
for i in range(915, 950):
    index_episode(i)
```

#### Upload the Wikipedia Page

```python
path = pathlib.Path(f"./ep_json/wiki.json")
payload = json.loads(path.read_text(encoding="utf-8"))
response = requests.post(url, headers=headers, json=payload)
response.text
```
